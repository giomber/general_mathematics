```python
import numpy as np
import matplotlib.pyplot as plt
```

This notebook is about gradient descent

```python
def f(x):
    return x**3-2*(x**2)+5
```

```python
print(f(3))
xx=np.linspace(0, 3, 100)
```

```python
plt.plot(np.linspace(0, 3, 100),f(np.linspace(0,3,100)))
```

```python
def grad_descent(x0, learning_rate=0.01, max_iter=10, tol=1e-6):
    x = x0
    iterations = np.array([x])
    for i in range(max_iter):
        grad = 3*x**2 - 4*x  # Derivative of f(x)
        x_new = x - learning_rate * grad
        iterations = np.append(iterations, x_new)
        if abs(x_new - x) < tol:
            break
        x = x_new
    return iterations
```

```python
algo=grad_descent(3, learning_rate=0.01, max_iter=20)
```

```python
plt.plot(np.linspace(0, 3, 100),f(np.linspace(0,3,100)))
plt.plot(algo,f(algo),'o')
plt.show()
```
